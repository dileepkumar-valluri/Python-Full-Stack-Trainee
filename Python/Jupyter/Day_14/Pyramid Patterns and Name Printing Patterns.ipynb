### Pyramid Patters

```python
n=int(input())
for i in range(n):
    for j in range(n-i):
        print('*',end=' ')
    print()
```

```python
n=int(input())
for i in range(n):
    #spaces
    for j in range(n-i):
        print('@',end=' ')
    #Stars
    for j in range(2*i+1):
        print('*',end=' ')
    print()
```

```python
n=int(input())
for i in range(n):
    #spaces
    for j in range(n-i):
        print(' ',end=' ')
    #Stars
    for j in range(2*i+1):
        print('*',end=' ')
    print()
```

```python
n=int(input())
for i in range(n):
    #spaces
    for j in range(n-1-i):
        print(' ',end=' ')
    #Stars
    for j in range(2*i+1):
        print('*',end=' ')
    print()
```

```python
n=int(input())
for r in range(n):
    #spaces
    for c in range(r):
        print(' ',end=' ')
    #stars
    for c in range(2*n-(2*r+1)):
        print('*',end=' ')
    print()
        
```

```python
n=int(input())
for i in range(n):
    #spaces
    for j in range(n-1-i):
        print(' ',end=' ')
    #Stars
    for j in range(2*i+1):
        print('*',end=' ')
    print()
for r in range(n):
    #spaces
    for c in range(r):
        print(' ',end=' ')
    #stars
    for c in range(2*n-(2*r+1)):
        print('*',end=' ')
    print()
```

```python
# Same Input value
n=int(input())
for i in range(n):
    #spaces
    for j in range(n-1-i):
        print(' ',end=' ')
    #Stars
    for j in range(2*i+1):
        print('*',end=' ')
    print()
for r in range(1,n):
    #spaces
    for c in range(r):
        print(' ',end=' ')
    #stars
    for c in range(2*n-(2*r+1)):
        print('*',end=' ')
    print()
```

```python
# Different Input Values
n=3
for i in range(n):
    #spaces
    for j in range(n-i-1):
        print(' ',end=' ')
    #Stars
    for j in range(2*i+1):
        print('*',end=' ')
    print()
n=2
for r in range(n):
    #spaces
    for c in range(r+1):
        print(' ',end=' ')
    #stars
    for c in range(2*n-(2*r+1)):
        print('*',end=' ')
    print()
```

```python
a=1
n=3
for i in range(n):
    #spaces
    for j in range(n-i-1):
        print(' ',end=' ')
    #Stars
    for j in range(2*i+1):
        print(a,end=' ')
        a+=1
    print()
```

### Name Printing Patterns

```python
name=input()
for ch in name:
    print(ch)
```

```python
name=input()
for ch in range(len(name)):
    print(name[:ch+1])
```

```python
name=input()
a=1
for ch in name:
    print(ch*a)
    a+=1
```

```python
ch=65 
print(chr(ch))
print(chr(ch+10))
```

```python
print(ord('a'))
print(ord('A'))
```

```python
name=input()
a=1
for ch in range(len(name)):
    print((name[ch])*a)
    a+=1
```

```python
m=[[10,20,20,40],[50,60,70,80],[90,100,200,300],[400,500,600,700]]
d1,d2=0,0
for i in range(len(m)):
    for j in range(len(m[0])):
        if i==j:
            d1+=m[i][j]
        elif (i+j)==(len(m)-1):
            d2+=m[i][j]
print(d1,d2)
print(abs(d1-d2))
```
