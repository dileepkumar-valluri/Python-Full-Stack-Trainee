### Scope

```python
# 1.Local Scope
def f():
    x=10
    print(x)
f()
```

```python
# 2.Global Scope
x=20
def f2():
    print(x)
f2()
print(x)
```

```python
# 3.Enclosing Scope
def outer():
    x='outer'
    def inner():
        y='inner'
        print(x,y)
    inner()
outer()
```

```python
# 4. Build in Function
def f3():
    l=[1,2,3]
    return sum(l)
f3()
```

```python
x=20
def f2():
    global x
    x=x+10
f2()
print(x)
```

```python
def outer():
    x=10
    def inner():
        nonlocal x
        x=x+100
        print(x)
    inner()
outer()
```

```python
# Pass by value
def f(x):
    x=20
    print(id(x))
    print(x)
a=10
print(id(a))
f(a)
```

```python
# Pass by Object reference
def f(x):
    x=[10,20,30]
    print(id(x))
    print(x)
a=[1,2,3,4]
f(a)
```

```python
# Pass by Object reference using immutable data type
def f(x):
    x=20
    print(id(x))
    print(x)
a=10
print(id(a))
f(a)
```

```python
# Pass by Object reference using immutable data type - modification or reassign
def f(x):
    x.append(10)
    print(id(x))
    x=[10,20,30,40]
    print(id(x))
    print(x)
a=[1,2,3,4]
print(id(a))
f(a)
```

### Recursive Function

```python
def factorial(n):
    if n==1 or n==0:
        return 1
    return n*factorial(n-1)
a=int(input())
print(factorial(a))
```
