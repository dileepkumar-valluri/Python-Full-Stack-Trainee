### Scope

```python
# 1.Local Scope
def f():
    x=10
    print(x)
f()
```

```python
# 2.Global Scope
x=20
def f2():
    print(x)
f2()
print(x)
```

```python
# 3.Enclosing Scope
def outer():
    x='outer'
    def inner():
        y='inner'
        print(x,y)
    inner()
outer()
```

```python
# 4. Build in Function
def f3():
    l=[1,2,3]
    return sum(l)
f3()
```

```python
x=20
def f2():
    global x
    x=x+10
f2()
print(x)
```

```python
def outer():
    x=10
    def inner():
        nonlocal x
        x=x+100
        print(x)
    inner()
outer()
```

```python
# Pass by value
def f(x):
    x=20
    print(id(x))
    print(x)
a=10
print(id(a))
f(a)
```

```python
# Pass by Object reference
def f(x):
    x=[10,20,30]
    print(id(x))
    print(x)
a=[1,2,3,4]
f(a)
```

```python
# Pass by Object reference using immutable data type
def f(x):
    x=20
    print(id(x))
    print(x)
a=10
print(id(a))
f(a)
```

```python
# Pass by Object reference using immutable data type - modification or reassign
def f(x):
    x.append(10)
    print(id(x))
    x=[10,20,30,40]
    print(id(x))
    print(x)
a=[1,2,3,4]
print(id(a))
f(a)
```

### Recursive Function

```python
# Finding Factorial of a numbers
def factorial(n):
    if n==1 or n==0:
        return 1
    return n*factorial(n-1)
a=int(input())
print(factorial(a))
```

```python
# Sum of N natural numbers
def natural_numbers(n):
    if n==1:
        return 1
    return n+natural_numbers(n-1)
print(natural_numbers(5))
```

```python
# Sum of N even numbers
def even_numbers(n):
    if n%2==0:
        if n==2:
            return 2
        return n+even_numbers(n-2)
    else:
        return even_numbers(n-1)
print(even_numbers(10))
```

```python
# Sum of N odd numbers
def even_numbers(n):
    if n%2!=0:
        if n==1:
            return 1
        return n+even_numbers(n-2)
    else:
        return even_numbers(n-1)
print(even_numbers(11))
```

```python
# Checking Prime number or not
def Prime_number(n):
    c=0
    if n%Prime_number(n-1)!=0:
        c+=1
    if c==2 and n==1:
        return True 
    else:
        return False
print(Prime_number(8))
```

```python
def Prime_Number(n):
    c=0
    for i in range(1,n+1):
        if n%i==0:
            c+=1
    if c==2:
        return True
    else:
        return False
n=int(input())
print(Prime_Number(n))
```

```python
# Sum of N Prime Numbers
def Sum_pn(n):
    l=[]
    for i in range(2,n+1):
        c=0
        for j in range(1,i+1):
            if i%j==0:
                c+=1
        if c==2:
            l.append(i)
    return l
n=int(input())
Sum_pn(n)
    
```
