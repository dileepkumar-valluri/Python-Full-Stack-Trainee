### Comparision Operator

```python
a=26
b=45
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

### Logical Operators

```python
a=26
b=45
print(a>b and a<b)
print(a<=b and a!=b)
print(a==b and b>=a)
print()
print(a>b or a<b)
print(a<=b or a!=b)
print(a==b or b>=a)
print()
print(not a>b)
print(not a<b)
```

### Bitwise Operator

```python
print(6 & 4)
print(6 | 4)
print(6^4)
print(4<<1)
print(8>>1)
print(8>>)
```

### Identity operator

```python
l1=[10,20,30]
l2=[10,20,30]
l3=l1
print(l1 is l2)
print(l1 is l3)
print(l1==l3)
```

### Membership Operator

```python
l1=[10,20,30]
l2=[10,20,30]
l3=l1
print(l1 in l2)
print(l1 in l3)
print(l1==l3)
print(id(l1),id(l2),id(l3))
```

### Problem

```python
length=int(input('Enter Length:'))
breadth=int(input('Enter breadth:'))
Area=length*breadth
Perimeter=2*(length+breadth)
print(f'Area of rectangle: {Area} having length {length} & breadth {breadth}')
print(f'Perimeter of rectange is {Perimeter} having length {length} and breadth {breadth}')
```
