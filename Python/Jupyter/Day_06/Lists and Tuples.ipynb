### List Methods

```python
l=[1,2,3,4,5,6,6,7]
print(l)
l.append(10)
print(l)
l.append([20,30])
print(l)
```

```python
l.extend([40,50,60])
print(l)
```

```python
l.insert(2,[70,80])
print(l)
l.insert(0,0)
print(l)
```

```python
print(l.pop())
```

```python
lst=[70,20,90,30,100,50]
print(len(lst))
print(min(lst))
print(max(lst))
print(sum(lst))
lst.sort()
print(lst)
print(sorted(lst))
lst.sort(reverse=True)
print(lst)
```
