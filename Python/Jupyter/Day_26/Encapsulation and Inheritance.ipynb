### Encapsulation

```python
class Company:
    def __init__(self,emp_name,emp_id,emp_salary):
        self.emp_name=emp_name
        self._emp_id=emp_id
        self.__emp_salary=emp_salary
    def employee_salary(self):
        return self.__emp_salary
    def emp_details(self):
        print(f' Employee Name is {self.emp_name} and Employee Id is {self._emp_id}')
emp_name=input()
emp_id=int(input())
emp_salary=int(input())
employee=Company(emp_name,emp_id,emp_salary)
print(employee.emp_name)
print(employee._emp_id)
employee.employee_salary()
print(employee.employee_salary())
```

### Inheritance

#### 1. Single Inheritance

```python
class Company:
    def main_agenda(self):
        print('Abstract of Comapny')
    def clients(self):
        print('Clients')
    def CEO(self):
        print('I am the CEO')
class Cloud_Company(Company):
    def cloud(self):
        print('Company Data in the Cloud')
c=Cloud_Company()
c.main_agenda()
c.cloud()
```

```python
class Employee:
    def employee_company(self):
        print('Employee Company')
    def employee_experience(self):
        print('20 years Experience')
class Location(Employee):
    def Living_location(self):
        print('Lives in Hyderabad')
    def Company_location(self):
        print('Company Location is Hyderabad')
l=Location()
l.employee_company()
l.employee_experience()
l.Company_location()
```
