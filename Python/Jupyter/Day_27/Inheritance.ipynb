#### 1.Single Inheritance

```python
class employee:
    def work(self):
        print('working')
class Developer(employee):
    def develop(self):
        print('Develop the Website')
d=Developer()
d.develop()
d.work()
```

#### 2.Multiple Inheritance 

```python
class Employee:
    def work(self):
        print('Working')
class Developer:
    def develop(self):
        print('Developing the website')
class tester(Employee,Developer):
    def Testing(self):
        print('Testing the Website')
t=tester()
t.work()
t.develop()
t.Testing()
```

#### 3.Multi Level Inheritance

```python
class Employee:
    def work(self):
        print('Working')
class Developer(Employee):
    def develop(self):
        print('Developing the website')
class tester(Developer):
    def Testing(self):
        print('Testing the Website')
t=tester()
t.work()
t.develop()
t.Testing()
```

#### 4. Hierarchical Inheritance

```python
class Company:
    def Rules(self):
        print('Follow the company rules')
class HR_team(Company):
    def Hiring(self):
        print('Hiring the Employees')
class frontend_team(Company):
    def Design(self):
        print('Designing the Website')
class Backend_Team(Company):
    def python_operations(self):
        print('Writing python code to develop the company')
hr=HR_team()
f=frontend_team()
b=Backend_Team()
hr.Rules()
hr.Hiring()
print()
f.Rules()
f.Design()
print()
b.Rules()
b.python_operations()
```

#### 5. Hybrid Inheritance

```python
class Company:
    def Rules(self):
        print('Follow the company rules')
class HR_team(Company):
    def Hiring(self):
        print('Hiring the Employees')
class frontend_team(Company):
    def Design(self):
        print('Designing the Website')
class Backend_Team(Company): #Hierarchical Inheritance
    def python_operations(self):
        print('Writing python code to develop the company')
class clients(HR_team,Backend_Team,Company): # Multiple Inheritance
    def meetups(self):
        print('client meetings')
client=clients()
client.meetups()
client.python_operations()
client.Hiring()
client.Rules()
```

```python
class Company:
    def Rules(self):
        print('Follow the company rules')
class HR_team:
    def Hiring(self):
        print('Hiring the Employees')
class frontend_team:
    def Design(self):
        print('Designing the Website')
class clients(Company,HR_team,frontend_team): # Multiple Inheritance
    def meetups(self):
        print('client meetings')
class Backend_team(clients): # Single Inheritance
    def python_operations(self):
        print('Writing python code to develop the company')
b=Backend_team()
b.python_operations()
b.meetups()
b.Design()
b.Hiring()
b.Rules()
```

```python
class A:
    def AA(self):
        print('class A')
class B(A): # Single Inheritance
    def BB(self):
        print('class B')
class C:
    def CC(self):
        print('class C')
class D(C,B): #Multiple Inheritance
    def DD(self):
        print('Class D')
class E(D): # Multi level Inheritance
    def EE(self):
        print('Class E')
class F(E):
    def FF(self):
        print('Class F')
class G(F): # Hierarchical Inheritance
    def GG(self):
        print('Class G')
class H(F):
    def HH(self):
        print('Class H')
class I(F):
    def II(self):
        print('Class I')

g=G()
g.FF()
g.EE()
g.DD()
g.CC()
g.BB()
g.AA()
print()
h=H()
h.FF()
h.EE()
h.DD()
h.CC()
h.BB()
h.AA()
print()
i=I()
i.FF()
i.EE()
i.DD()
i.CC()
i.BB()
i.AA()
```

#### Super() method

```python
class A:
    def method1(self):
        print('This is parent method1')
class B(A):
    def method1(self):
        super().method1()
        print('this is child method1')
b=B()
b.method1()
```

```python
class A:
    def __init__(self,name,roll_num):
        self.name=name
        self.roll_num=roll_num
    def method1(self):
        print('This is parent method1')
class B(A):
    def __init__(self,name,roll_num,marks):
        super().__init__(name,roll_num)
        self.marks=marks
    def method1(self):
        super().method1()
        print('this is child method1')
        print(f'Name is {self.name}, Roll Number is {self.roll_num} and marks are {self.marks}')
name=input()
roll_num=int(input())
marks=int(input())
b=B(name,roll_num,marks)
b.method1()
```
