### 2.Random

```python
import random
random.randint(1,6)
n=[10,20,40,30,60]
print(random.choice(n))
print(random.choices(n,k=3))
print(random.sample(n,k=2))
print(random.random())
print(random.randrange(1,6))
print(random.uniform(1,6))
random.shuffle(n)
print(n)
```

### 5. Collections

```python
from collections import Counter
name='codegnan'
r=Counter(name)
print(r)
```

```python
from collections import Counter
nums=[1,2,3,4,1,2,10,2,2]
n=Counter(nums)
print(n)
```

```python
name=input()
d={}
for ch in name:
    if ch not in d:
        d[ch]=1
    else:
        d[ch]+=1
print(d)
```

```python
from collections import defaultdict
d=defaultdict(int)
d['apple']=1
print(d['name'])
print(d)
dic={'a':2,'b':5,'c':10}
# print(dic['x']) # error due to x element is not in dict
```

```python
from collections import deque
d=deque([1,2,3])
print(d)
d.append(4)
print(d)
d.appendleft(10)
print(d)
d.pop()
print(d)
d.popleft()
print(d)
```

### 6. Itertools

```python
from itertools import permutations
numbers=[1,2,3,4]
print(list(permutations(numbers)))
```

```python
from itertools import combinations
print(list(combinations(numbers,2)))
```

```python
from itertools import product
A=[1,2,4]
B=['A','B','C']
print(list(product(A,B)))
```

### 7.Datetime

```python
from datetime import datetime
now=datetime.now()
print(now)
print()
print(now.year)
print(now.month)
print(now.day)
print()
print(now.hour)
print(now.minute)
print(now.second)
```

```python
from datetime import date
today=date.today()
print(today)
# Manula Creation of a date
d=date(2026,10,10)
print(d)
```

```python
from datetime import date,timedelta
today=date.today()
future=today+timedelta(days=30)
print(future)
past=today-timedelta(days=5)
print(past)
```

```python
from datetime import datetime
now=datetime.now()
print(now.strftime('%d-%m-%y'))
print(now.strftime('%d-%m-%Y'))
```
