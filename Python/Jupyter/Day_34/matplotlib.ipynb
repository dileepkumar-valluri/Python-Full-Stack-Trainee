### Line Plot

```python
import matplotlib.pyplot as mplt
months=['Jan','Feb','Mar','Apr']
sales=[25,32,28,47]
mplt.plot(months,sales)
mplt.show()
```

```python
import matplotlib.pyplot as mplt
months=['Jan','Feb','Mar','Apr']
sales=[25,32,28,47]
mplt.plot(months,sales,marker='o')
mplt.show()
```

```python
import matplotlib.pyplot as mplt
months=['Jan','Feb','Mar','Apr']
sales=[25,32,28,47]
mplt.plot(months,sales,marker='o',color='red')
mplt.grid()
mplt.show()
```

```python
import matplotlib.pyplot as mplt
months=['Jan','Feb','Mar','Apr']
sales=[25,32,28,47]
mplt.plot(months,sales,marker='o',color='red')
mplt.title('Monthly Sales')
mplt.xlabel('Months')
mplt.ylabel('Sales')
mplt.grid()
mplt.show()
```

#### Bar Graph

```python
subjects=['Python','DSA','Aptitude','Sofskills']
marks=[89,85,70,60]
mplt.bar(subjects,marks)
mplt.title('Students Marks Report')
mplt.xlabel('Subjects')
mplt.ylabel('Marks')
mplt.show()
```

```python
subjects=['Python','DSA','Aptitude','Sofskills']
marks=[89,85,70,60]
mplt.barh(subjects,marks)
mplt.title('Students Marks Report')
mplt.xlabel('Subjects')
mplt.ylabel('Marks')
mplt.show()
```

#### Pie Chart

```python
subjects=['Python','DSA','Aptitude','Sofskills']
marks=[89,85,70,60]
mplt.pie(marks,labels=subjects,startangle=90,autopct='%1.1f%%')
mplt.title('Students Marks Report')
mplt.show()
```

#### Histogram

```python
# Histogram shows the distribution of numerical data
import matplotlib.pyplot as plt
marks=[35,40,42,45,48,50,52,55,57,60,62,65,68,70,72]
mplt.hist(marks,bins=3)
mplt.title('Distribution of Marks')
mplt.xlabel('Marks')
mplt.ylabel('Number of Students')
mplt.show()
```

#### Scatter Plot

```python
# Represents individual data points
# It is mainly used to understand the relationship between two numerical variable
# import matplotlob.pyplot as plt

hours=[1,2,3,4,5,6]
marks=[35,40,50,60,70,85]
mplt.scatter(hours,marks)
mplt.title('Study Hours vs Marks')
mplt.grid()
mplt.xlabel('Study Hours')
mplt.ylabel('Marks')
plt.show()
```

#### Subplot

```python
fig,ax=mplt.subplots(2,3,figsize=(18,6))
ax[0,0].plot([1,2,3],[10,20,15])
ax[0,0].set_title('Line')

ax[0,1].bar(['A','B','C'],[10,20,15])
ax[0,1].set_title('Bar')

ax[1,0].scatter([1,2,3],[10,20,15])
ax[1,0].set_title('Scatter')

ax[1,1].plot([1,2,3],[5,15,10])
ax[1,1].set_title('Another Line')

ax[0,2].bar(['A','B','C'],[10,20,15])
ax[0,2].set_title('Bar')

ax[1,2].scatter([1,2,3],[10,20,15])
ax[1,2].set_title('Scatter')

mplt.show()
```

```python
import matplotlib.pyplot as mplt
fig,ax=mplt.subplots(2,3,figsize=(18,6))
ax[0,0].plot([1,2,3],[10,20,15])
ax[0,0].set_title('Line')

ax[0,1].barh(['A','B','C'],[10,20,15])
ax[0,1].set_title('Bar')

ax[1,0].scatter([1,2,3],[10,20,15])
ax[1,0].set_title('Scatter')

ax[1,1].pie([1,2,3])
ax[1,1].set_title('Pie Chart')

ax[0,2].hist(['A','B','C'],bins=2)
ax[0,2].set_title('Histogram')

ax[1,2].boxplot([1,2,3])
ax[1,2].set_title('Box Plot')

mplt.show()
```
