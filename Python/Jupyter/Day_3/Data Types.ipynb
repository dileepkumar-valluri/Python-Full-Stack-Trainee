### Numeric Data Types

```python
number=26
cgpa=8.9
c=2+8j
print(type(number),type(cgpa),type(c))
```

```python
c=5+6j
print(c)
```

### Sequence data type

```python
#List
lst=list()
lst1=[]
lst2=[1,2,3,5]
lst3=[10,'jhfrew',67.88,'h']
print(type(lst1))
print(type(lst2))
lst3[1]=45
print(type(lst3))
print(lst2[3])
print(lst3)
```

```python
#Tuples
t=()
t1=tuple()
t2=(1,1,2,4,5)
print(tuple(t),tuple(t1),tuple(t2))
print(t2[3])
```

```python
#Strings
string='jfakdsnj@#$%^&*'
s=''
s1=str()
print(type(string),type(s),type(s1))
print(s)
print(string[5])
string[5]=23
print(string) # Error because string is Immutale
```

### Mapping Data Type

```python
#Sets & Dictionaries
set1={}
dictionary={}
set2=set()
set3={1,2,3,4,5,6,11,1,1,1,1}
d1=dict()
d2={}
print(d2)
print(d1)
d={'name':'Dileep','age':21,'Gender':'Male','CGPA':8.71,9:'i'}
for key,value in d.items():
    print(key,value)
print(d.keys())
print(d.items())
print(type(set1),type(dictionary),type(set2),type(set3),type(d1),type(d1))
print(set3)
```

```python
d2={'a':1,'b':1}
print(d2)
```

```python
d3={'a':1,'b':1,'a':4}
print(d3)
```

```python
#Type Casting

#Implicit Type Casting
print(True+6)
print(float(9))
print(4+9.4)
```

```python
#Explicit Type Casting
str1='lkvf'
l=list(str1)
print(l)
```
