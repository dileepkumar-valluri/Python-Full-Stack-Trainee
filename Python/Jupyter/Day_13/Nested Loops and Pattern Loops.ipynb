```python
n=int(input())
for row in range(n):
    for column in range(n):
        print('*',end=' ')
    print()
```

```python
n=int(input())
for r in range(n):
    for c in range(n):
        if r==c or (r+c)==n-1:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
n=int(input())
for i in range(1,n+1):
    for j in range(1,n+1):
        if i==(n//2)+1 or j==(n//2)+1:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
n=int(input())
for i in range(n):
    for j in range(n):
        if (i+j)%2==0:
            print('1',end=' ')
        else:
            print('0',end=' ')
    print()
```

```python
n=int(input())
even=2
odd=1
for i in range(n):
    for j in range(n):
        if i%2==0:
            print(even,end=' ')
            even+=2
        else:
            print(odd,end=' ')
            odd+=1
    print()
```
