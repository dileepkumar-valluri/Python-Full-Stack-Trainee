### Regualar Expression Methods

```python
import re
text = 'Python Programming'
print(re.match('Python',text))
print(re.match('Programming',text))
```

```python
print(re.search('Programming',text))
```

### Email Validation

```python
import re
user1_Email=input()
pattern=r'\w+@\w+\.\w+'
if re.fullmatch(pattern,user1_Email):
    print('Valid')
else:
    print('Invalid')
```

```python
import re
user1_Email=input()
pattern=r'^\w+@\w+\.\w+$'
if re.search(pattern,user1_Email):
    print('Valid')
else:
    print('Invalid')
```

### Phone Number Validation

```python
import re
num=input()
ph=r'\d{10}'
if re.fullmatch(ph,num):
    print('Valid')
else:
    print('Invalid')
```

```python
import re
num=input()
ph=r'[6-9]\d{9}'
if re.fullmatch(ph,num):
    print('Valid')
else:
    print('Invalid')
```

```python
import re
num=input()
ph=r'\+91[6-9]\d{9}'
if re.fullmatch(ph,num):
    print('Valid')
else:
    print('Invalid')
```

```python
import re
num='ph1 +919234567890 ph2 +1234567890 and ph3 +911234567890'
ph=r'\+91[6-9]\d{9}'
p=re.findall(ph,num)
print(p)
```

### Credit Card Number Validation

```python
import re
num=input()
c=r'\d{4}\s\d{4}\s\d{4}\s\d{4}'
if re.fullmatch(c,num):
    print('Valid')
else:
    print('Invalid')
```

### Date Validation

```python
import re
date=input()
d=r'\d{2}-\d{2}-\d{4}'
if re.fullmatch(d,date):
    print('Valid')
else:
    print('Invalid')
```

### Employee Validation

```python
import re
emp=input()
e=r'CGTD\d{4}'
if re.fullmatch(e,emp):
    print('Valid')
else:
    print('Invalid')
```

### URL Validation

```python
import re
url=input()
u=r'https?://\w+\.\w+\.\w+'
if re.fullmatch(u,url):
    print('Valid')
else:
    print('Invalid')
```

### Password Validation

```python
import re
pw=input()
password=r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"
if re.fullmatch(password,pw):
    print('Valid')
else:
    print('Invalid')
```
