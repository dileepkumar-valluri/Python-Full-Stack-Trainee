```python
square=lambda x: x*x
print(square(20))
```

```python
# Greater element
greater= lambda a,b: a if a>b else b
print(greater(70,28))
```

```python
# Filter()
numbers=[1,2,3,4,5,6]
res=filter(lambda x: str(x), numbers)
print(res)
```

```python
numbers=[1,2,3,4,5,6]
res=filter(lambda x: str(x), numbers)
print(res)
print(list(res))
```

```python
numbers=[1,2,3,4,5,6]
res=filter(lambda x: x%2==0, numbers)
print(res)
print(list(res))
```

```python
# map()
numbers=map(int,input().split())
print(list(numbers))
```

```python
# reduce()
from functools import reduce
numbers=[1,2,3,4,5]
res=reduce(lambda a,b: a+b,numbers)
print(res)
```
