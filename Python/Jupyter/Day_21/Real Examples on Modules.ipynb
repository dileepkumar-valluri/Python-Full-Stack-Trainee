```python
import random
characters='ABCDEFGHIJKLMNOPQRSTUVWXYZ'
characters+='abcdefghijklmnopqrstuvwxyz'
characters+='0123456789'
characters+='!@#$%^&*'
length=int(input())
password=''
for i in range(length):
    ch=random.choice(characters)
    password+=ch
print(password)
```

```python
import random
upper_case='ABCDEFGHIJKLMNOPQRSTUVWXYZ'
lower_case='abcdefghijklmnopqrstuvwxyz'
numbers='0123456789'
symbols='!@#$%^&*'
password=[]
length=int(input())
if length<4:
    print(f'your password length is {length}, choose password length higher')
else:
    password.append(random.choice(upper_case))
    password.append(random.choice(lower_case))
    password.append(random.choice(numbers))
    password.append(random.choice(symbols))
for i in range(length-4):
    all_characters=upper_case+lower_case+numbers+symbols
    ch=random.choice(all_characters)
    password.append(ch)
    random.shuffle(password)
new_password=''.join(password)
print(new_password)
```

```python
import time
seconds=0
minutes=0
hours=0
while True:
    print(f'{hours}:{minutes}:{seconds}')
    time.sleep(1)
    seconds+=1
    if seconds==60:
        minutes+=1
        seconds=0
        if minutes==60:
            hours+=1
            minutes=0
```

```python
import time
from IPython.display import clear_output
seconds=0
minutes=0
hours=0
while True:
    print(f'{hours}:{minutes}:{seconds}')
    time.sleep(1)
    seconds+=1
    if seconds==60:
        minutes+=1
        seconds=0
        if minutes==60:
            hours+=1
            minutes=0
    clear_output(wait=True)
```

```python
import random
rolls=0
count=0
while True:
    dice=random.randint(1,6)
    if count+dice<100:
        count+=dice
    rolls+=1
    print(f'You are at {dice+count} on {rolls} rolls')
    if count==100:
        print(f'You won the match on {rolls}')
        break
```
