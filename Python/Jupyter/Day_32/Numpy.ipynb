#### Mathematical Operations

#### Scalar operation

```python
a=np.array([10,20,30])
b=np.array([1,2,3])
print(a+b)
print(a+5)
print(a-1)
print(b*3)
```

#### Array operations with Aggregate Functions

```python
arr=np.array([10,20,30,40,50])
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.sum(arr))
```
