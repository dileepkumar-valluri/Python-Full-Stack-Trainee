```python
class DSA:
    def __init__(self,name,ID):
        self.name=name
        self.ID=ID
    def student_details(self,marks):
        self.marks=marks
        print(f'Student name is {self.name}, student ID is {self.ID} and got the marks {marks}')
    def Problem_solving(self):
        print('Problem Solving')
    def listening(self):
        print('Listening')
name=input()
ID=input()
PFS65=DSA(name,ID)
PFS66=DSA(name,ID)
PFS66.listening()
PFS65.Problem_solving()
PFS66.student_details(100)
```
