```python
### Standardize Mobile Number without Using Decorators
n=int(input())
for i in range(n):
    num=int(input())
    phno=str(num)
    if len(phno)>10:
        if phno[0]=='0':
            print('+91 '+phno[1:6]+' '+phno[6:])
        elif phno[0:2]=='91':
            print('+91 '+phno[2:7]+' '+phno[7:])
        elif phno[0:3]=='+91':
            print('+91 '+phno[3:8]+' '+phno[8:])
    else:
        print('+91 '+phno[0:5]+' '+phno[5:])
```
